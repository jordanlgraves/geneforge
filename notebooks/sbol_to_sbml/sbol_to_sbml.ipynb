```python
base_api_url = 'http://localhost:4000'

#Parameter	Type	Description
#sbol	file	SBOL file to convert (.sbol or .xml)
#topModelId	string	Default: topModel
#What to name the top model file in the case of multiple output files.
#📞 callback	URL	Callback URL
# All parameters are should be provided in the body of the request as multipart/form-data
import requests

sbol_file = '../GeneticToggleSwitch/GeneticToggleSwitch.rdf'
topModelId = 'topModel'
# callback = 'http://localhost:4000/callback'

response = requests.post(base_api_url + '/sync/convert', files={'sbol': open(sbol_file, 'rb')})
```

```python
# Write to response.text to a file
with open('response_sbml.xml', 'w') as f:
    f.write(response.text)
```
